## Part 1: Spanning Probability $\Pi(p; L)$

### Theory

In **1D site percolation**, each site is independently occupied with probability $p$.

A **spanning cluster** exists only if **ALL** $L$ sites are occupied (since any gap breaks the connection).

Therefore, the spanning probability is:

$$\boxed{\Pi(p; L) = p^L}$$

This is simply the probability that $L$ independent events (each with probability $p$) all occur.

```python
# Generate probability values
p = np.linspace(0, 1, 500)

# Lattice sizes to plot
L_values = [L2, L1, L3]  # L = 1, 5, 13

# Create figure
plt.figure(figsize=(10, 7))

colors = ['blue', 'green', 'red']
for i, L in enumerate(L_values):
    Pi = p ** L  # Spanning probability: Π(p; L) = p^L
    plt.plot(p, Pi, color=colors[i], linewidth=2.5, label=f'L = {L}')

# Mark critical point
plt.axvline(x=1, color='black', linestyle='--', linewidth=1.5, alpha=0.7, label=r'$p_c = 1$')

# Labels and formatting
plt.xlabel('Occupation Probability $p$', fontsize=12)
plt.ylabel(r'Spanning Probability $\Pi(p; L)$', fontsize=12)
plt.title(r'Spanning Probability in 1D Percolation' + '\n' + r'$\Pi(p; L) = p^L$', fontsize=14)
plt.legend(fontsize=11, loc='upper left')
plt.grid(True, alpha=0.3)
plt.xlim(0, 1.05)
plt.ylim(0, 1.05)

# Annotation
plt.annotate('Curves become steeper\nas L increases', 
             xy=(0.85, 0.3), fontsize=10, style='italic',
             bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.savefig('figures/spanning_probability.png', dpi=150, bbox_inches='tight')
plt.show()

print("\nFigure saved: figures/spanning_probability.png")
```

### Interpretation of Spanning Probability Curves

**Observations from the figure:**

1. **L = 1 (Blue curve):** $\Pi(p; 1) = p$ — Linear relationship. Any single occupied site trivially "spans" a lattice of size 1.

2. **L = 5 (Green curve):** $\Pi(p; 5) = p^5$ — Moderately steep. All 5 sites must be occupied, so spanning probability drops significantly for $p < 1$.

3. **L = 13 (Red curve):** $\Pi(p; 13) = p^{13}$ — Very steep near $p = 1$. Even at $p = 0.9$, we have $\Pi \approx 0.25$.

**Key observation:** As $L$ increases, the curves become **steeper** and approach a **step function** at $p = 1$.

### Why Critical Probability $p_c = 1$

In the **thermodynamic limit** ($L \to \infty$):

$$\Pi(p; L) = p^L \xrightarrow{L \to \infty} 
\begin{cases} 
0 & \text{if } p < 1 \\
1 & \text{if } p = 1
\end{cases}$$

**Physical interpretation:** In 1D, there are **no alternative paths**. A single unoccupied site breaks the spanning cluster. Therefore, spanning is possible **only** when $p = 1$ (all sites occupied).

$$\boxed{p_c = 1 \text{ for 1D percolation}}$$

## Part 2: Average Cluster Size $S(p)$

### Theory

The **average cluster size** (excluding the infinite cluster) for 1D percolation is given by the exact formula:

$$\boxed{S(p) = \frac{1 + p}{1 - p}} \quad \text{for } p < 1$$

**Derivation insight:** This formula comes from summing over all possible finite cluster sizes weighted by their probability.

### Critical Behavior

Near $p_c = 1$:

$$S(p) = \frac{1 + p}{1 - p} \approx \frac{2}{1 - p} = \frac{2}{p_c - p}$$

Thus, $S(p) \sim (p_c - p)^{-\gamma}$ with **critical exponent** $\gamma = 1$.

```python
# Domain restriction: p < 1 (up to 0.99 to avoid division by zero)
p_cluster = np.linspace(0, 0.99, 500)

# Exact formula for average cluster size
S_p = (1 + p_cluster) / (1 - p_cluster)

# x-axis: (p_c - p) = (1 - p)
x_axis = 1 - p_cluster  # Ranges from 1 to 0.01

# Create figure
plt.figure(figsize=(10, 7))

plt.plot(x_axis, S_p, 'b-', linewidth=2.5, label=r'$S(p) = \frac{1+p}{1-p}$')

# Labels and formatting
plt.xlabel(r'$(p_c - p) = (1 - p)$', fontsize=12)
plt.ylabel(r'Average Cluster Size $S(p)$', fontsize=12)
plt.title(f'Average Cluster Size vs $(p_c - p)$ in 1D Percolation (L = {L4})\n' + 
          r'Exact Formula: $S(p) = \frac{1+p}{1-p}$', fontsize=14)
plt.legend(fontsize=11, loc='upper right')
plt.grid(True, alpha=0.3)

# Annotation for divergence
plt.annotate(r'$S(p) \rightarrow \infty$ as $p \rightarrow p_c = 1$' + '\n(Divergence at critical point)', 
             xy=(0.1, S_p[int(0.9*len(S_p))]), 
             xytext=(0.3, S_p[int(0.9*len(S_p))]*0.7),
             fontsize=10, style='italic',
             arrowprops=dict(arrowstyle='->', color='red'),
             bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# Invert x-axis so it goes from 1 (far from p_c) to 0 (at p_c)
plt.gca().invert_xaxis()

plt.tight_layout()
plt.savefig('figures/average_cluster_size.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"\nFigure saved: figures/average_cluster_size.png")
```

### Interpretation of Average Cluster Size Curve

**Observations from the figure:**

1. **At low $p$ (far from $p_c$):** $S(p) \approx 1$ — Clusters are small and isolated.

2. **As $p$ increases:** Clusters merge and grow larger on average.

3. **As $p \to p_c = 1$:** $S(p) \to \infty$ — The average cluster size **diverges**.

**Critical behavior:**

$$S(p) \sim \frac{1}{p_c - p} = \frac{1}{1-p} \quad \text{as } p \to 1$$

This divergence is a **hallmark of critical phenomena** and indicates the emergence of an infinite spanning cluster at $p_c$.

The **critical exponent** $\gamma = 1$ characterizes this power-law divergence.

---

## Key Conclusions

### 1. Critical Probability
$$p_c = 1 \text{ for 1D percolation}$$

### 2. Spanning Probability
$$\Pi(p; L) = p^L \xrightarrow{L \to \infty} \text{Step function at } p = 1$$

### 3. Average Cluster Size
$$S(p) = \frac{1+p}{1-p} \sim \frac{1}{p_c - p} \quad (\gamma = 1)$$

### 4. Fundamental Difference from Higher Dimensions

> **Unlike higher dimensions, 1D percolation does NOT exhibit a phase transition at finite $p$; criticality occurs ONLY at $p = 1$.**

This is because in 1D, any single unoccupied site completely breaks the spanning cluster. In 2D and 3D, alternative paths exist, allowing percolation at $p_c < 1$ (e.g., $p_c \approx 0.593$ for 2D square lattice).
